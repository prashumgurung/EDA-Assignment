We have age in numerical format, we need in categorical format( in groups)

```python
df['Age'].describe()
```

min value = 18
max value = 85

divide them in groups of:
10-19
20-29
30-39
40-49
50-59
60-69
70-79
80-89


```python
def age_categories(age):
    if (age >=10) & (age <20):
        return '< 20'
    elif (age >=20) & (age < 30):
        return '20 - 29'
    elif (age >= 30) & (age < 40):
        return '30 - 39' 
    elif (age >= 40) & (age < 50) :
        return '40 - 49'
    elif (age>= 50) & (age< 60):
        return '50 - 59'
    elif (age >= 60) & (age < 70):
        return '60 - 69'
    elif (age >= 70) & (age < 80):
        return '70 - 79'
    elif (age>=80) & (age < 90):
        return '> 80'
    else:
       return 'No Category'
```

```python
df['Age Caategories']=df['Age'].apply(age_categories)
```

```python
sns.barplot(data=df_age,x='Age Caategories',y='Billing Amount',palette='Set2')
plt.title('Age X Billing Amount')
plt.show()
```

```python
age_order=['< 20','20 - 29','30 - 39','40 - 49','50 - 59','60 - 69','70 - 79','> 80']
sns.barplot(data=df_age,x='Age Caategories',y='Billing Amount',palette='Set2',order=age_order)
plt.title('Age X Billing Amount')
plt.show()
```
